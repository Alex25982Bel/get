```python
# 4-1.py
import RPi.GPIO as GPIO

def convert_to(value):
    return list(int(x) for x in bin(value)[2:].zfill(8))

GPIO.setwarnings(False)
dac = [8, 11, 7, 1, 0, 5, 12, 6]
GPIO.setmode(GPIO.BCM)
GPIO.setup(dac, GPIO.OUT)

try:
    while True:
        number = input("Type a number from 0 to 255: ")
        if number == "q":
            break
        try:
            number = int(number)
            if 0 <= number <= 255:
                GPIO.output(dac, convert_to(number))
                voltage = float(number) / 256.0 * 3.3
                print(f"Output voltage is about {voltage:.4f} volt")
            else:
                if number < 0:
                    print("Number have to be >= 0! Try again...")
                else:
                    print("Number is out of range [0, 255]! Try again...")
        except ValueError:
            print("You have to type a number, not a symbol. Try again...")
finally:
    GPIO.output(dac, 0)
    GPIO.cleanup()
    print('EOP')
```

```python
# 4-2.py
import RPi.GPIO as GPIO
from time import sleep

def decimal2binary(value):
    return [int(element) for element in bin(value)[2:].zfill(8)]

dac = [8, 11, 7, 1, 0, 5, 12, 6]  
GPIO.setmode(GPIO.BCM)
GPIO.setup(dac, GPIO.OUT)

inc_flag = 1
x = 0

try:
    period = float(input("Type a period for signal: "))
    if period <= 0:
        raise ValueError("Period must be positive")

    while True:
        GPIO.output(dac, decimal2binary(x))
        voltage = float(x) / 256.0 * 3.3
        print(f"Output voltage is about {voltage:.4f} volt")

        if x == 0:
            inc_flag = 1
        elif x == 255:
            inc_flag = 0

        x = x + 1 if inc_flag == 1 else x - 1
        sleep(period / 512)

except ValueError:
    print("Inappropriate period")
finally:
    GPIO.output(dac, 0)
    GPIO.cleanup()
    print('EOP')
```

```python
# 4-3.py
import RPi.GPIO as GPIO

GPIO.setmode(GPIO.BCM)
GPIO.setup(27, GPIO.OUT)

p = GPIO.PWM(27, 1000)  
p.start(0)  

try:
    while True:
        try:
            f = int(input("Введите значение (0-100): "))
            if 0 <= f <= 100:
                p.ChangeDutyCycle(f)
                print(f"Выходное напряжение: {3.3 * f / 100:.2f} В")
            else:
                print("Значение должно быть от 0 до 100!")
        except ValueError:
            print("Ошибка: введите целое число!")

finally:
    p.stop()
    GPIO.output(27, 0)
    GPIO.cleanup()
```
