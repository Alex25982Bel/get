```python
# 5-1.py
import RPi.GPIO as GPIO
import time

dac = [8, 11, 7, 1, 0, 5, 12, 6]  
comp = 14
troyka = 13
levels = 2 ** 8  

GPIO.setmode(GPIO.BCM) 

GPIO.setup(dac, GPIO.OUT, initial=GPIO.LOW)  
GPIO.setup(troyka, GPIO.OUT, initial=GPIO.HIGH)
GPIO.setup(comp, GPIO.IN)

def convert_to(value):
    return [int(x) for x in bin(value)[2:].zfill(8)]  

def adc():
    for value in range(256):
        signal = convert_to(value)
        GPIO.output(dac, signal)
        time.sleep(0.0007)
        if GPIO.input(comp) == 0:  
            return value
    return 255

try:
    while True:
        signal = adc()
        voltage = signal / levels * 3.3
        print(f"Number: {signal}, Voltage: {voltage:.2f} V")  
        time.sleep(0.01)

except KeyboardInterrupt:
    print("\nThe program was stopped by the keyboard")
else:
    print("No exceptions")

finally:
    GPIO.output(dac, GPIO.LOW)
    GPIO.output(troyka, GPIO.LOW)
    GPIO.cleanup()
    print("GPIO cleanup completed")
```

```python
# 5-2.py
import RPi.GPIO as GPIO
import time

dac = [8, 11, 7, 1, 0, 5, 12, 6]  
comp = 14
troyka = 13
levels = 2 ** 8  

GPIO.setmode(GPIO.BCM)

GPIO.setup(dac, GPIO.OUT, initial=GPIO.LOW)  
GPIO.setup(troyka, GPIO.OUT, initial=GPIO.HIGH)  
GPIO.setup(comp, GPIO.IN)

def convert_to(value):
    return [int(x) for x in bin(value)[2:].zfill(8)]  

def adc():
    value = 0
    for i in range(7, -1, -1):
        new_value = value | (1 << i)  
        GPIO.output(dac, convert_to(new_value))
        time.sleep(0.0007)
        if GPIO.input(comp) == 0:
            value = new_value
    return value

try:
    while True:
        start = time.time()
        signal = adc()
        print(f"Conversion time: {time.time() - start:.4f} sec")
        voltage = signal / levels * 3.3
        print(f"Number: {signal}, Voltage: {voltage:.2f} V")  
        time.sleep(0.01)

except KeyboardInterrupt:
    print("\nThe program was stopped by the keyboard")
else:
    print("No exceptions")

finally: 
    GPIO.output(dac, GPIO.LOW)
    GPIO.output(troyka, GPIO.LOW)
    GPIO.cleanup()
    print("GPIO cleanup completed")
```

```python
# 5-3.py
import RPi.GPIO as GPIO
import time

dac = [8, 11, 7, 1, 0, 5, 12, 6]  
comp = 14
troyka = 13
levels = 256  
leds = [2, 3, 4, 17, 27, 22, 10, 9]  

GPIO.setmode(GPIO.BCM)

GPIO.setup(dac, GPIO.OUT, initial=GPIO.LOW)
GPIO.setup(leds, GPIO.OUT, initial=GPIO.LOW)
GPIO.setup(troyka, GPIO.OUT, initial=GPIO.HIGH)
GPIO.setup(comp, GPIO.IN)  

def convert_to(value):
    return [int(x) for x in bin(value)[2:].zfill(8)]  

def adc():
    value = 0
    for i in range(7, -1, -1):
        new_value = value | (1 << i)  
        GPIO.output(dac, convert_to(new_value))
        time.sleep(0.0007)
        if GPIO.input(comp) == 0:
            value = new_value
    return value

try:
    while True:
        signal = adc()
        voltage = signal / levels * 3.3
        print(f"Number: {signal}, Voltage: {voltage:.2f} V")  
        leds_count = min(8, max(0, round(signal / 32)))  
        leds_state = [1] * leds_count + [0] * (8 - leds_count)
        GPIO.output(leds, leds_state)
        time.sleep(0.01)

except KeyboardInterrupt:
    print("\nThe program was stopped by the keyboard") 
else:
    print("No exceptions")

finally:  
    GPIO.output(dac, GPIO.LOW)
    GPIO.output(troyka, GPIO.LOW)
    GPIO.cleanup()
```
